```python
# 0. 라이브러리 불러오기
import pandas as pd

# 1. 데이터 경로 설정
BUSINESS_JSON_PATH = '../data/yelp_academic_dataset_business.json'

# 2. 데이터 불러오기
def load_business_data(path):
    print(f"Loading business data from {path}")
    return pd.read_json(path, lines=True)

# 3. 음식점 데이터 필터링
def filter_restaurants(df):
    restaurant_df = df[df['categories'].str.contains('Restaurants', na=False)].copy()
    print(f"Filtered {len(restaurant_df)} restaurant entries")
    return restaurant_df

# 4. 도시 기준 필터링 (Las Vegas, Phoenix)
def filter_by_city(df, cities):
    cities = [city.lower().strip() for city in cities]
    df['city_clean'] = df['city'].str.lower().str.strip()
    filtered = df[df['city_clean'].isin(cities)].copy()
    print(f"Filtered {len(filtered)} entries for cities: {cities}")
    return filtered

# 5. 주요 컬럼만 추출
def simplify_columns(df):
    simple_df = df[[
        'business_id', 'name', 'city', 'state', 'stars', 'review_count'
    ]].copy()
    print(f"Simplified data shape: {simple_df.shape}")
    return simple_df

# 6. 실행
business_df = load_business_data(BUSINESS_JSON_PATH)
restaurant_df = filter_restaurants(business_df)
# selected_cities = ['Las Vegas', 'Phoenix']
selected_cities = ['las vegas', 'phoenix']
filtered_df = filter_by_city(restaurant_df, selected_cities)
simple_df = simplify_columns(filtered_df)

# 테스트 
# 1. 도시 이름 확인
# print(business_df['city'].unique())

# # 2. 음식점 필터링 후 일부 데이터 확인
# print(restaurant_df.head())

# # 3. 필터링 후 도시 확인
# print(filtered_df['city'].unique())

# # 7. 결과 확인
# print(simple_df['city'].value_counts())
# simple_df.head(3)

filtered_cities = business_df[business_df['city'].str.contains('las vegas|phoenix', case=False, na=False)]
print(filtered_cities['city'].unique())

# 2. 필터링된 데이터 출력
print(filtered_cities.head())
```
