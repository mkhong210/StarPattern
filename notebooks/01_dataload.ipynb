```python
# data load 최종버전

# 0. 라이브러리 불러오기
import pandas as pd
import os

# 1. 데이터 경로 설정
BUSINESS_JSON_PATH = '../data/yelp_academic_dataset_business.json'
REVIEW_JSON_PATH = '../data/yelp_academic_dataset_review.json'

# 2. 비즈니스 데이터 불러오기
def load_business_data(path):
    """business.json 불러오기"""
    print(f"Loading business data from {path}")
    return pd.read_json(path, lines=True)

# 3. 음식점만 필터링
def filter_restaurants(df):
    """카테고리 중 'Restaurants' 포함된 데이터만 추출"""
    restaurant_df = df[df['categories'].str.contains('Restaurants', na=False)].copy()
    print(f"Filtered {len(restaurant_df)} restaurant entries")
    return restaurant_df

# 4. 필요한 컬럼만 추출
def simplify_restaurant_data(df):
    """business_id, name, city, state, stars, review_count만 추출"""
    simple_df = df[[ 'business_id', 'name', 'city', 'state', 'stars', 'review_count' ]].copy()
    print(f"Simplified data shape: {simple_df.shape}")
    return simple_df

# 5. 리뷰 데이터 불러오기
def load_review_data(path):
    """review.json 불러오기"""
    print(f"Loading review data from {path}")
    return pd.read_json(path, lines=True)

# 6. 음식점 리뷰만 필터링
def filter_restaurant_reviews(review_df, restaurant_df):
    """business_id 기준으로 음식점 리뷰만 추출"""
    restaurant_ids = set(restaurant_df['business_id'])
    filtered_reviews = review_df[review_df['business_id'].isin(restaurant_ids)].copy()
    print(f"Filtered {len(filtered_reviews)} restaurant-related reviews")
    return filtered_reviews

# 7. 실행
business_df = load_business_data(BUSINESS_JSON_PATH)
restaurant_df = filter_restaurants(business_df)
restaurant_simple_df = simplify_restaurant_data(restaurant_df)

review_df = load_review_data(REVIEW_JSON_PATH)
restaurant_reviews = filter_restaurant_reviews(review_df, restaurant_simple_df)

# 8. 저장
restaurant_simple_df.to_csv('../data/restaurant_business.csv', index=False)
restaurant_reviews.to_csv('../data/restaurant_reviews.csv', index=False)

# 9. 샘플 확인
print(restaurant_simple_df.head())
print(restaurant_reviews.head())
```
