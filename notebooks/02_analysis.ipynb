```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 불러오기
reviews = pd.read_csv('../data/restaurant_reviews.csv')

# 리뷰 길이 컬럼 추가
reviews['review_length'] = reviews['text'].apply(len)

# 시각화
sns.scatterplot(data=reviews.sample(10000), x='review_length', y='stars', alpha=0.3)
plt.title('리뷰 길이와 별점의 관계')
plt.xlabel('리뷰 길이 (문자 수)')
plt.ylabel('별점')
plt.show()

# 히트맵으로 상관관계 시각화
# plt.figure(figsize=(8, 6))
# sns.heatmap(correlation, annot=True, cmap='coolwarm', vmin=-1, vmax=1)
# plt.title('별점과 리뷰 길이의 상관계수 히트맵')
# plt.show()

# 박스플롯롯
plt.figure(figsize=(8, 6))
sns.boxplot(data=reviews, x='stars', y='review_length')
plt.title('별점별 리뷰 길이 분포')
plt.xlabel('별점')
plt.ylabel('리뷰 길이 (문자 수)')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 불러오기
reviews = pd.read_csv('../data/restaurant_reviews.csv')

# 리뷰 길이 컬럼 추가
reviews['review_length'] = reviews['text'].apply(len)

# 바이올린플롯
plt.figure(figsize=(8, 6))
sns.violinplot(data=reviews, x='stars', y='review_length', inner='box', density_norm='width')
plt.title('별점별 리뷰 길이 분포 (Violin Plot)')
plt.xlabel('별점')
plt.ylabel('리뷰 길이 (문자 수)')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 별점별 평균 리뷰 길이 계산
mean_lengths = reviews.groupby('stars')['review_length'].mean().reset_index()

# 시각화
plt.figure(figsize=(8, 5))
sns.barplot(data=mean_lengths, x='stars', y='review_length', hue='stars', palette='Blues_d', legend=False)
plt.title('별점별 평균 리뷰 길이')
plt.xlabel('별점')
plt.ylabel('평균 리뷰 길이 (문자 수)')
plt.show()
```
